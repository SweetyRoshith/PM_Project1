VALIDATING DATE FIELDS

```python
###CHECKING ALL THE DATATYPES
df_col_types= df.dtypes.astype(str).to_dict()
print("Columns with their data types\n",df_col_types) #list of all columns with their datatypes- as a dictionary.

#FINDING DATE COLUMNS,THEIR EXISTING DATATYPE,IDENTIFYING THE FORMAT OF THE VALUES(if DD/MM/YYYY or MM/DD/YYYY)
#  & parsing to DATE data type in the identified format

for column,dtype in df_col_types.items():
    if 'date'in column.lower().strip():   ##identifying date columns
        date_string=df[column].str.strip()
        # CHECKING FOR DATE FORMAT  DD/MM/YYYY or MM/DD/YYYY & Parsing datatype to Date in the identified format
        date_parts = date_string.str.split('/', expand=True)
        day = pd.to_numeric(date_parts[0], errors='coerce')
        month = pd.to_numeric(date_parts[1], errors='coerce')
        # print(date_parts[2])
        dd_mm_format = day > 12
        dd_mm_count = dd_mm_format.sum()
        mm_dd_format = month > 12
        mm_dd_count = mm_dd_format.sum()

        if dd_mm_count>0 & mm_dd_count==0:
            print("******IDENTIFIED DATE FORMAT IS DD/MM/YYYY HH:MM*******")
            # print("BEFORE",df[column])
            df[column] = pd.to_datetime(df[column], format='%d/%m/%Y %H:%M',errors='coerce')
            print("*****Column "+column+ " parsed to datatype DATE******")
        if mm_dd_count>0 & dd_mm_count==0:
            print("******IDENTIFIED DATE FORMAT IS MM/DD/YYYY HH:MM******") 
            # print("BEFORE",df[column])
            df[column] = pd.to_datetime(df[column], format='%m/%d/%Y %H:%M',errors='coerce')
            print("*****Column "+column+ " parsed to datatype DATE******")
# print(df.dtypes)
```
