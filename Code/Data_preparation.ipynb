LOADING THE DATA INTO A DATAFRAME

```python
import pandas as pd

##Fixing UnicodeDecodeError in position 1709
##Inspecting for the actual encoding of the Source file as it failed to load with the default encoding
import chardet
with open(r"../data/raw/DataCoSupplyChainDataset.csv","rb") as f:
    raw=f.read(100000)
result=chardet.detect(raw)

##Using the encoding detected
df=pd.read_csv(r"../data/raw/DataCoSupplyChainDataset.csv",encoding=result['encoding'])
```

MISSING VALUES CHECK and DUPLICATES CHECK

```python
#MISSING VALUES CHECK
missing=df.isnull().sum() #counts the null values against each column
missing=missing[missing>0].reset_index() #creating a dataframe of only the columns with missing values
missing.columns=["Column","Missing Values"] #assigning column headers
missing["Missing Percentage"] = (missing["Missing Values"] / len(df) * 100).round(4) #percentage of missing values columnwise
display(missing)
```

```python
#DUPLICATES CHECK
df.duplicated().sum()
```
