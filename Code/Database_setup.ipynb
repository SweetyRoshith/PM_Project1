Loading the data in csv into SQLite database

```python
#importing necessary packages
import pandas as pd
import sqlite3

#reading the existing CSV
df=pd.read_csv(r"..\data\raw\DataCoSupplyChainDataset.csv",encoding="Windows-1252") #Your file path here
#display(df)

#creates a folder 'database',if already doesn't exist, in the root folder PM_Project1.
os.makedirs(r"..\database", exist_ok=True)
# Create/connect to a new .db file — this creates the file if it doesn't exist
conn=sqlite3.connect(r"..\database\PM_Project1.db")
df.to_sql("original_data_table",conn,if_exists="replace",index=False)
conn.close()
```
